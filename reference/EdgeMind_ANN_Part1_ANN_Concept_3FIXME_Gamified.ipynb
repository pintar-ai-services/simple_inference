## Training Mission

Now the neural network is ready to learn.

The training cycle is:

```text
Clear Gradients → Forward Pass → Calculate Loss → Backpropagation → Optimizer Step
```

During training:

- The model makes a prediction.
- The loss function measures the error.
- Backpropagation calculates gradients.
- The optimizer updates the weights.
- The prediction improves gradually.

````python
# ============================================================
# SECTION 7: TRAIN THE MODEL
# ============================================================

epochs = 50
training_loss = []

for epoch in range(epochs):

    optimizer.zero_grad()
    prediction = model(x)
    loss = criterion(prediction, target)
    loss.backward()
    optimizer.step()

    training_loss.append(loss.item())

    if (epoch + 1) % 10 == 0:
        print("Epoch:", epoch + 1, "/", epochs, "| Loss:", round(loss.item(), 6))

print("\nTraining completed!")
````

## Mission Observation — Before vs After Training

The goal is to observe that the model has learned by reducing the loss.

````python
# ============================================================
# SECTION 8: COMPARE BEFORE AND AFTER TRAINING
# ============================================================

prediction_after = model(x)
loss_after = criterion(prediction_after, target)

print("====================================")
print("MISSION OBSERVATION")
print("====================================")
print("Prediction before training:", round(prediction_before.item(), 6))
print("Prediction after training :", round(prediction_after.item(), 6))
print("\nTarget value              :", target.item())
print("\nLoss before training      :", round(loss_before.item(), 6))
print("Loss after training       :", round(loss_after.item(), 6))

if loss_after.item() < loss_before.item():
    print("\n✅ The loss decreased. The ANN has learned!")
else:
    print("\n⚠️ The loss did not decrease. Try running the training again or check the FIX ME answers.")
````

## Export for deployment — two files

Complete the three FIX ME tasks and run all training and evaluation cells first. Then run the export cell below; no extra exercise or manual JSON editing is needed.

It creates `deployment/ann_concept/` with:

- **`model.pt`** — the trained weights, saved on CPU for portability.
- **`model.json`** — the architecture, input settings, preprocessing, output interpretation, labels where applicable, and an example input.

This introductory model predicts a number from three numeric inputs. It was trained on just one sample, so it demonstrates deployment mechanics rather than useful generalization. It has no class labels or scaler.

Copy **both files** into the same `models/ann_concept/` folder in the deployment repository. The JSON always refers to its neighboring `model.pt`; do not mix files from different exports. Add the folder name to the `models` list in `config/ann.json`, then restart both the API and inference worker and reload the ANN Inference tab. PIMA is the included default; your registered model becomes another choice in the model selector.

The export checks that its predictions match the trained model before writing the files. If you retrain or change the model, run this cell again and copy both new files.


````python
# SECTION 11: EXPORT FOR DEPLOYMENT

# Run this after completing the FIX ME tasks, training, and evaluation.
import copy
import io
import json
from pathlib import Path

MODEL_ID = "ann_concept"
DISPLAY_NAME = "ANN Concept"

# The introductory notebook already uses nn.Sequential.
deployment_model = copy.deepcopy(model).cpu().eval()
check_input = x.detach().cpu()
input_config = {
    "type": "tabular", "dtype": "float32", "shape": [None, 3],
    "features": ["feature_1", "feature_2", "feature_3"],
    "preprocessing": {"type": "none"},
}
output_config = {
    "task": "regression", "shape": [None, 1],
    "target": "target", "labels": [], "postprocessing": {"type": "none"},
}
example = {"features": dict(zip(input_config["features"], check_input[0].tolist()))}

# Describe the exact layer order so serving does not need this notebook's class.
architecture = []
for layer in deployment_model:
    if isinstance(layer, nn.Linear):
        architecture.append({
            "type": "linear", "in_features": layer.in_features,
            "out_features": layer.out_features, "bias": layer.bias is not None,
        })
    elif isinstance(layer, nn.ReLU):
        architecture.append({"type": "relu"})
    elif isinstance(layer, nn.Flatten):
        architecture.append({
            "type": "flatten", "start_dim": layer.start_dim, "end_dim": layer.end_dim,
        })
    else:
        raise ValueError(f"Export does not support layer: {type(layer).__name__}")

config = {
    "schema_version": 1,
    "model_id": MODEL_ID,
    "display_name": DISPLAY_NAME,
    "model_version": "1",
    "framework": "pytorch",
    "pytorch_version": str(torch.__version__),
    "weights": {"file": "model.pt", "format": "state_dict"},
    "architecture": {"type": "sequential", "layers": architecture},
    "input": input_config,
    "output": output_config,
    "example": example,
}

# Check that the portable model and saved weights preserve the original outputs.
# Moving copies to CPU leaves the trained model on its original device.
original_cpu = copy.deepcopy(model).cpu().eval()
deployment_model.eval()
weights_buffer = io.BytesIO()
torch.save(deployment_model.state_dict(), weights_buffer)
weights_buffer.seek(0)
reloaded_model = copy.deepcopy(deployment_model)
reloaded_model.load_state_dict(torch.load(weights_buffer, map_location="cpu", weights_only=True))
with torch.inference_mode():
    expected = original_cpu(check_input)
    actual = reloaded_model(check_input)
torch.testing.assert_close(actual, expected)

# The folder contains the two files students copy together for deployment.
export_dir = Path("deployment") / MODEL_ID
export_dir.mkdir(parents=True, exist_ok=True)
config_text = json.dumps(config, indent=2, allow_nan=False) + "\n"
(export_dir / "model.pt").write_bytes(weights_buffer.getvalue())
(export_dir / "model.json").write_text(config_text, encoding="utf-8")

print("Export verified: predictions match the trained model.")
print("Export folder:", export_dir.resolve())
print("Copy BOTH model.pt and model.json into models/" + MODEL_ID + "/.")
print("Keep these files together; export both again after retraining.")
````
