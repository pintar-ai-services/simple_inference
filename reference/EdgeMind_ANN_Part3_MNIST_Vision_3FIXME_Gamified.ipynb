````python
# ============================================================
# SECTION 1: IMPORT LIBRARIES AND CHECK JETSON DEVICE
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import DataLoader
from torchvision import datasets, transforms

torch.manual_seed(42)
np.random.seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("====================================")
print("EDGEMIND AI: MNIST VISION")
print("====================================")
print("PyTorch version:", torch.__version__)
print("Running device:", device)

if device.type == "cuda":
    print("CUDA is available. Jetson GPU can be used.")
else:
    print("CUDA is not available. Notebook will run on CPU.")
````

## Dataset Briefing — MNIST Handwritten Digits

MNIST contains handwritten digit images.

Each image is:

```text
28 pixels × 28 pixels
```

A basic ANN does not process the 2D image directly.  
The image is flattened into:

```text
784 input values
```

### Image Classification Concept

```text
Pixel Values → ANN Model → Digit Class
```

````python
# ============================================================
# SECTION 2: LOAD MNIST DATASET
# ============================================================

transform = transforms.ToTensor()

train_dataset = datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

test_dataset = datasets.MNIST(
    root="./data",
    train=False,
    download=True,
    transform=transform
)

print("MNIST dataset loaded successfully!")
print("Training samples:", len(train_dataset))
print("Testing samples:", len(test_dataset))
````

````python
# ============================================================
# SECTION 3: CREATE DATA LOADERS
# ============================================================

batch_size = 64

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False,
    num_workers=0
)

print("DataLoaders are ready!")
print("Batch size:", batch_size)
````

## Training Mission

Now the neural vision model is ready to learn.

Training cycle:

```text
Forward Pass → Calculate Loss → Backpropagation → Optimizer Step
```

You already activated:

```text
Network Representation → Loss Function → Optimization
```

````python
# ============================================================
# SECTION 9: TRAIN THE MODEL
# ============================================================

epochs = 3
training_loss = []

for epoch in range(epochs):

    model.train()
    running_loss = 0.0

    for batch_images, batch_labels in train_loader:

        batch_images = batch_images.to(device)
        batch_labels = batch_labels.to(device)

        optimizer.zero_grad()

        outputs = model(batch_images)

        loss = criterion(outputs, batch_labels)

        loss.backward()

        optimizer.step()

        running_loss += loss.item()

    average_loss = running_loss / len(train_loader)
    training_loss.append(average_loss)

    print("Epoch:", epoch + 1, "/", epochs, "| Loss:", round(average_loss, 4))

print("\nTraining completed on:", device)
````

````python
# ============================================================
# SECTION 10: VISUALIZE TRAINING LOSS
# ============================================================

plt.figure(figsize=(8, 5))
plt.plot(training_loss, marker="o", label="Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("EdgeMind AI: MNIST Vision Training Loss")
plt.legend()
plt.grid(True)
plt.show()
````

## Evaluation Mission

The model is tested using unseen digit images.

For multi-class classification:

```text
Class Scores → Highest Score → Predicted Digit
```

````python
# ============================================================
# SECTION 11: EVALUATE MODEL
# ============================================================

model.eval()

correct = 0
total = 0

with torch.no_grad():

    for batch_images, batch_labels in test_loader:

        batch_images = batch_images.to(device)
        batch_labels = batch_labels.to(device)

        outputs = model(batch_images)

        _, predicted = torch.max(outputs, 1)

        total += batch_labels.size(0)
        correct += (predicted == batch_labels).sum().item()

accuracy = correct / total

print("====================================")
print("MNIST VISION RESULT")
print("====================================")
print("Test Accuracy:", round(accuracy * 100, 2), "%")

if accuracy >= 0.95:
    print("\n🏆 BONUS ACHIEVED!")
    print("Neural Vision Builder Plus unlocked!")
elif accuracy >= 0.90:
    print("\n✅ MISSION PASSED!")
    print("Badge Unlocked: Neural Vision Builder")
else:
    print("\n⚠️ MISSION NOT COMPLETED YET")
    print("Try increasing epochs, adjusting hidden neurons or learning rate.")
````

## Export for deployment — two files

Complete the three FIX ME tasks and run all training and evaluation cells first. Then run the export cell below; no extra exercise or manual JSON editing is needed.

It creates `deployment/mnist/` with:

- **`model.pt`** — the trained weights, saved on CPU for portability.
- **`model.json`** — the architecture, input settings, preprocessing, output interpretation, labels where applicable, and an example input.

The JSON describes one-channel 28×28 images with pixel values scaled from 0–255 to 0–1, matching the notebook's `ToTensor()` transform. Use a light digit on a dark background, as in MNIST. Flattening is included in the exported architecture; serving must not flatten the image a second time. The labels are digits 0–9 in output-index order.

Copy **both files** into the same `models/mnist/` folder in the deployment repository. The JSON always refers to its neighboring `model.pt`; do not mix files from different exports. Add the folder name to the `models` list in `config/ann.json`, then restart both the API and inference worker and reload the ANN Inference tab. PIMA is the included default; your registered model becomes another choice in the model selector.

The export checks that its predictions match the trained model before writing the files. If you retrain or change the model, run this cell again and copy both new files.


````python
# SECTION 13: EXPORT FOR DEPLOYMENT

# Run this after completing the FIX ME tasks, training, and evaluation.
import copy
import io
import json
from pathlib import Path

MODEL_ID = "mnist"
DISPLAY_NAME = "MNIST Vision"

# Include the flatten operation from forward() in the deployment architecture.
deployment_model = nn.Sequential(
    nn.Flatten(start_dim=1), copy.deepcopy(model.fc1), nn.ReLU(),
    copy.deepcopy(model.fc2), nn.ReLU(), copy.deepcopy(model.fc3),
).cpu().eval()
check_input = sample_images[:1].detach().cpu()
input_config = {
    "type": "image", "dtype": "float32", "shape": [None, 1, 28, 28],
    "layout": "NCHW",
    "preprocessing": {
        "type": "image", "color_mode": "L", "resize": [28, 28],
        "resize_interpolation": "bilinear", "pixel_scale": 1.0 / 255.0,
        "invert": False, "background": "black", "foreground": "white",
    },
}
output_config = {
    "task": "multiclass_classification", "shape": [None, 10], "target": "digit",
    "labels": [str(digit) for digit in range(10)],
    "postprocessing": {"type": "softmax", "dim": 1},
}
# Embed an example image in JSON so deployment still needs only two files.
example = {
    "image": {
        "encoding": "uint8_pixels", "shape": [28, 28],
        "pixels": (check_input[0, 0] * 255).round().to(torch.uint8).tolist(),
    }
}

# Describe the exact layer order so serving does not need this notebook's class.
architecture = []
for layer in deployment_model:
    if isinstance(layer, nn.Linear):
        architecture.append({
            "type": "linear", "in_features": layer.in_features,
            "out_features": layer.out_features, "bias": layer.bias is not None,
        })
    elif isinstance(layer, nn.ReLU):
        architecture.append({"type": "relu"})
    elif isinstance(layer, nn.Flatten):
        architecture.append({
            "type": "flatten", "start_dim": layer.start_dim, "end_dim": layer.end_dim,
        })
    else:
        raise ValueError(f"Export does not support layer: {type(layer).__name__}")

config = {
    "schema_version": 1,
    "model_id": MODEL_ID,
    "display_name": DISPLAY_NAME,
    "model_version": "1",
    "framework": "pytorch",
    "pytorch_version": str(torch.__version__),
    "weights": {"file": "model.pt", "format": "state_dict"},
    "architecture": {"type": "sequential", "layers": architecture},
    "input": input_config,
    "output": output_config,
    "example": example,
}

# Check that the portable model and saved weights preserve the original outputs.
# Moving copies to CPU leaves the trained model on its original device.
original_cpu = copy.deepcopy(model).cpu().eval()
deployment_model.eval()
weights_buffer = io.BytesIO()
torch.save(deployment_model.state_dict(), weights_buffer)
weights_buffer.seek(0)
reloaded_model = copy.deepcopy(deployment_model)
reloaded_model.load_state_dict(torch.load(weights_buffer, map_location="cpu", weights_only=True))
with torch.inference_mode():
    expected = original_cpu(check_input)
    actual = reloaded_model(check_input)
torch.testing.assert_close(actual, expected)

# The folder contains the two files students copy together for deployment.
export_dir = Path("deployment") / MODEL_ID
export_dir.mkdir(parents=True, exist_ok=True)
config_text = json.dumps(config, indent=2, allow_nan=False) + "\n"
(export_dir / "model.pt").write_bytes(weights_buffer.getvalue())
(export_dir / "model.json").write_text(config_text, encoding="utf-8")

print("Export verified: predictions match the trained model.")
print("Export folder:", export_dir.resolve())
print("Copy BOTH model.pt and model.json into models/" + MODEL_ID + "/.")
print("Keep these files together; export both again after retraining.")
````
